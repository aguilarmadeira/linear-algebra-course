```python
import numpy as np
np.set_printoptions(precision=4, suppress=True)   # saída mais legível / cleaner output
```

**Produto interno e norma** (5.1, 5.2). / *Inner product and norm.*

```python
u = np.array([1, 2, -1])
v = np.array([3, 0, 2])
print('u . v =', u @ v)                          # 1
print('||(1,2,-2)|| =', np.linalg.norm([1, 2, -2]))  # 3
```

**Gram–Schmidt com matrizes: $A=QR$** (5.4). As colunas de $Q$ são a base ortonormada obtida à mão a partir de $(3,1)$ e $(2,2)$, a menos do sinal. / *Columns of Q: the orthonormal basis found by hand, up to sign.*

```python
A = np.array([[3, 2],
              [1, 2]])          # colunas (3,1) e (2,2)
Q, R = np.linalg.qr(A)
print('Q =\n', Q)
print('R =\n', R)                  # triangular superior / upper triangular
print('Q^T Q =\n', Q.T @ Q)        # identidade / identity
```

```python
print('à mão / by hand:', np.array([3, 1]) / np.sqrt(10), np.array([-1, 3]) / np.sqrt(10))
```

**Extensão: projeção sobre um subespaço** $V=L\{(1,1,0),(0,0,1)\}$. / *Extension: projection onto a subspace.*

```python
e1 = np.array([1, 1, 0]) / np.sqrt(2)
e2 = np.array([0, 0, 1])
w = np.array([3, 1, 2])
p = (w @ e1) * e1 + (w @ e2) * e2
print('proj_V w =', p, '  resto / remainder:', w - p)   # (2,2,2), (1,-1,0)
print('resto . e1 =', (w - p) @ e1, '  resto . e2 =', (w - p) @ e2)
```

**Confere com os slides.** / *Matches the slides.*
