```python
import numpy as np
np.set_printoptions(precision=4, suppress=True)   # saída mais legível / cleaner output
```

A matriz transversal da UC e o sistema $C\mathbf x=(2,-11,-9)$. / *The course's running matrix and the system.*

```python
C = np.array([[ 2, 0,  0],
              [-6, 2, -3],
              [-6, 0, -1]])
b = np.array([2, -11, -9])
```

**Característica** (3.3) — à mão obtivemos $\operatorname{car} C=3$. / *Rank — by hand we got 3.*

```python
np.linalg.matrix_rank(C)
```

**Inversa** (3.5) — compare com a obtida por Gauss–Jordan. / *Inverse — compare with Gauss–Jordan.*

```python
np.linalg.inv(C)
```

**Sistema** (3.4) — `solve` usa eliminação de Gauss **sem calcular a inversa**: em geral, é mais eficiente e numericamente mais estável. / *`solve` uses Gaussian elimination without computing the inverse.*

```python
x = np.linalg.solve(C, b)
print('x =', x)              # (1, 2, 3)
print('C x - b =', C @ x - b)
```
