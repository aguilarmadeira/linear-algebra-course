```python
import numpy as np
np.set_printoptions(precision=4, suppress=True)   # saída mais legível / cleaner output
import math
```

```python
C = np.array([[ 2, 0,  0],
              [-6, 2, -3],
              [-6, 0, -1]])
print('det C =', float(np.linalg.det(C)))   # -4, com erro de arredondamento / with rounding error
```

O computador **não usa a definição** (soma de $n!$ termos): condensa a matriz e multiplica os pivôs (4.2). Porquê? / *The computer does not use the definition. Why?*

```python
n = 20
termos = math.factorial(n)
anos = termos / 1e9 / (365.25 * 24 * 3600)
print(f'definição / definition: {termos:.2e} termos -> cerca de {anos:.0f} anos a 1e9 termos/s')
print(f'condensação / elimination: cerca de {n**3/3:.0f} operações')
```

**Cuidado:** não se testa `det(A) == 0`. `matrix_rank` usa uma tolerância para estimar numericamente a característica. / *Never test `det(A) == 0`; `matrix_rank` uses a tolerance.*

```python
print('det C + 4 =', np.linalg.det(C) + 4)   # ~1e-15 ou 0, conforme o computador / depends on the machine
print(np.isclose(np.linalg.det(C), -4))       # comparar com tolerância / compare with a tolerance
print(np.linalg.matrix_rank(C))               # 3
```
