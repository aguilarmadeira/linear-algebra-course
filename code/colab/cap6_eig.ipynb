**Exemplo 4**: à mão, $\lambda=2$ (vetores $(1,0,-2)$, $(0,1,0)$) e $\lambda=-1$ (vetor $(0,1,1)$). O computador devolve vetores **de norma 1**, por outra ordem: são os mesmos, a menos de um fator. / *Same eigenvectors up to a non-zero factor.*

```python
C = np.array([[ 2, 0,  0],
              [-6, 2, -3],
              [-6, 0, -1]], dtype=float)
lam, P = np.linalg.eig(C)
print('valores próprios / eigenvalues:', lam)
print('P =\n', P)
print('P^-1 C P =\n', np.linalg.inv(P) @ C @ P)   # diagonal
```

**Cidade e campo**: aplicar $M$ repetidamente leva $\mathbf x_k$ ao vetor próprio de $\lambda=1$, $(2/3,\,1/3)$ — o *método das potências*, ideia de base do PageRank. / *City and countryside: the power method.*

```python
M = np.array([[0.9, 0.2],
              [0.1, 0.8]])
x = np.array([0.5, 0.5])
for k in range(1, 51):
    x = M @ x
    if k in (1, 10, 50):
        print(f'x_{k} =', x)
```
